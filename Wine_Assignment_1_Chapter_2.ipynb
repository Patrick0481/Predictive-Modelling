```python
import matplotlib.pyplot as plt 

plt.rc('font', size=14)
plt.rc('axes', labelsize=14, titlesize=14)
plt.rc('legend', fontsize=14)
plt.rc('xtick', labelsize=10)
plt.rc('ytick', labelsize=10)

White_wine_df.hist(bins=20, figsize=(12, 10), layout=(4, 4))

plt.show()
```

```python
#total sulfur dioxide and volatile acidity
#x=White_wine_df[White_wine_df['quality'].isin([3, 4, 5, 6, 7, 8, 9])][['total sulfur dioxide']]
#y=White_wine_df[White_wine_df['quality'].isin([3, 4, 5, 6, 7, 8, 9])]['volatile acidity']

x=White_wine_df[White_wine_df['quality'].isin([3, 4, 5, 6, 7, 8, 9])][['density']]
y=White_wine_df[White_wine_df['quality'].isin([3, 4, 5, 6, 7, 8, 9])]['residual sugar']

plt.scatter(x, y, c=White_wine_df[White_wine_df['quality'].isin([3, 4, 5, 6, 7, 8, 9])]['quality'], label='Class 0')
plt.title('Features with a large impact')
plt.xlabel("density")
plt.ylabel('residual sugar')

plt.show()
```

```python

#No this data is not linearly separable, as we can see from the scatter plot above. The classes are not well separated and there is a lot of overlap between the classes. This means that a linear model will not be able to separate the classes well and will have a poor performance.
```

```python
#Yes, we can't use a linear model to classify this data, as the classes are not linearly separable. We can use a non-linear model such as a decision tree or a random forest to classify this data.
```

```python
import seaborn as sns
correlation_matrix = White_wine_df.corr()
fig, ax = plt.subplots(figsize=(12, 8))
cax = ax.matshow(correlation_matrix, cmap='coolwarm')
fig.colorbar(cax) 
plt.title("Correlation Matrix using matshow", pad=20)
plt.show()
```

```python
plt.figure(figsize=(10, 8))
correlation_matrix = White_wine_df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.1f', linewidths=0) 
plt.title("Correlation Matrix using seaborn")
plt.show()
```

```python
#Density and residual sugar have the highest correlation with quality
#White wine quality is positively correlated with density and residual sugar, which means that as the density and residual sugar increase, the quality of the wine also increases.
```

```python
#The data is not linearly separable with the two variables that have the highest correlation with quality, which means that we can't use a linear model to classify this data. 
```
